##1. Definisi dan Esensi Machine Learning

Machine Learning (ML) bukan sekadar pemrograman biasa. Secara teoritis, ML adalah bidang ilmu yang memberikan komputer kemampuan untuk belajar dari data guna menyelesaikan tugas tertentu tanpa instruksi eksplisit.

- Pendekatan Tradisional: Input + Aturan (Rules) $\rightarrow$ Output.

- Pendekatan ML: Input + Output $\rightarrow$ Aturan (Model).

- Kriteria Tom Mitchell: Sebuah program komputer dikatakan belajar dari Experience (E) sehubungan dengan Task (T) dan Performance (P) jika performanya pada T, yang diukur oleh P, meningkat seiring dengan bertambahnya E.

## 2. Mengapa Menggunakan Machine Learning?
ML sangat dibutuhkan untuk skenario berikut:

- Masalah Kompleks: Masalah yang memerlukan daftar aturan yang terlalu panjang untuk ditulis tangan (contoh: deteksi spam).

- Lingkungan Dinamis: Masalah yang polanya berubah seiring waktu sehingga sistem perlu beradaptasi secara otomatis.

- Insight (Data Mining): Menggali volume data besar untuk menemukan pola yang tidak disadari oleh manusia.

## 3. Jenis-Jenis Sistem Machine Learning
Sistem ML dikategorikan berdasarkan tiga kriteria utama:

A. Berdasarkan Pengawasan (Supervision)
Supervised Learning: Data latih menyertakan solusi (label). Contoh: Regresi dan Klasifikasi.

1. Unsupervised Learning: Data tidak memiliki label. Contoh: Clustering (pengelompokan), Visualization, dan Anomaly Detection.

2. Semisupervised Learning: Menggabungkan sedikit data berlabel dengan banyak data tidak berlabel.

3. Reinforcement Learning: Agen belajar melalui sistem reward dan penalty untuk mendapatkan kebijakan (policy) terbaik.

B. Berdasarkan Kemampuan Adaptasi
- Batch Learning: Model dilatih secara offline dengan semua data. Untuk data baru, model harus dilatih ulang dari awal.

- Online Learning: Model dilatih secara bertahap (incremental) dengan data yang masuk secara kontinu. Cocok untuk sistem dengan keterbatasan memori.

C. Berdasarkan Cara Generalisasi
- Instance-based Learning: Mempelajari data mentah dan membandingkan data baru menggunakan ukuran kemiripan (similarity).

- Model-based Learning: Membangun model matematika dari data untuk membuat prediksi.

## 4. Tantangan Utama Machine Learning
Keberhasilan ML tidak hanya soal algoritma, tapi juga kualitas data:

- Kuantitas Data Kurang: Algoritma butuh ribuan hingga jutaan sampel untuk belajar dengan baik.

- Data Tidak Representatif: Sampel data tidak mencerminkan populasi asli (Sampling Bias).

- Kualitas Buruk: Data penuh dengan noise, error, atau outlier.

- Fitur Tidak Relevan: Data mengandung informasi yang tidak membantu proses prediksi (butuh Feature Engineering).

- Overfitting: Model terlalu kompleks sehingga "menghafal" noise pada data latih dan gagal di data baru.

- Underfitting: Model terlalu sederhana untuk menangkap pola dasar data.

## 5. Testing dan Validasi
Untuk memastikan model bisa bekerja di dunia nyata (generalisasi), kita menggunakan:

- Training Set: Untuk melatih model.

## 2. Memahami Masalah dan Memilih Metrik Performa

Sebelum mulai melakukan *coding*, langkah penting yang harus dilakukan adalah memahami jenis permasalahan Machine Learning yang akan diselesaikan serta menentukan metrik performa yang tepat.

### 2.1 Menentukan Jenis Masalah

Jenis masalah Machine Learning umumnya dibagi menjadi:

- **Regression**  
  Digunakan ketika target berupa nilai numerik kontinu, seperti harga rumah atau suhu.

- **Classification**  
  Digunakan ketika target berupa kelas atau kategori, seperti spam/tidak spam.

### 2.2 Memilih Metrik Performa

Pemilihan metrik performa harus disesuaikan dengan karakteristik data dan tujuan bisnis.

- **Root Mean Square Error (RMSE)**  
  Merupakan metrik standar untuk tugas regresi yang memberikan penalti lebih besar pada kesalahan (error) yang besar.

  $$
  RMSE(X, h) = \sqrt{\frac{1}{m}\sum_{i=1}^{m}\left(h(x^{(i)}) - y^{(i)}\right)^2}
  $$

- **Mean Absolute Error (MAE)**  
  Digunakan ketika data memiliki banyak *outlier* (pencilan), karena lebih robust terhadap nilai ekstrem.

  $$
  MAE(X, h) = \frac{1}{m}\sum_{i=1}^{m}\left|h(x^{(i)}) - y^{(i)}\right|
  $$



## 3. Persiapan Data (Data Cleaning & Transformation)

Tahap persiapan data merupakan bagian yang paling memakan waktu, namun sangat krusial karena kualitas data sangat memengaruhi performa model Machine Learning.

### 3.1 Menangani Data yang Hilang (*Missing Data*)

Data yang hilang harus ditangani dengan tepat agar tidak menyebabkan bias atau error pada model. Beberapa pendekatan yang umum digunakan antara lain:

- Menghapus baris (row) atau entri data yang memiliki nilai hilang.
- Menghapus seluruh fitur (kolom) yang memiliki banyak data tidak lengkap.
- Mengisi nilai yang hilang (*imputation*) dengan nilai tertentu, seperti:
  - **Zero**
  - **Mean**
  - **Median**  
  Pengisian ini dapat dilakukan menggunakan `SimpleImputer`.

### 3.2 Menangani Teks dan Atribut Kategorikal

Sebagian besar algoritma Machine Learning hanya dapat memproses data numerik. Oleh karena itu, data kategorikal perlu diubah menjadi bentuk angka.

- **Ordinal Encoding**  
  Digunakan untuk kategori yang memiliki urutan atau tingkatan.  
  Contoh: `Low`, `Medium`, `High`.

- **One-Hot Encoding**  
  Digunakan untuk kategori yang tidak memiliki urutan.  
  Contoh: nama kota atau jenis warna.

### 3.3 Penskalaan Fitur (*Feature Scaling*)

Penskalaan fitur diperlukan agar algoritma tidak bias terhadap fitur dengan rentang nilai yang lebih besar (misalnya gaji dibandingkan umur).

- **Min-Max Scaling (Normalization)**  
  Mengubah nilai fitur ke dalam rentang 0 hingga 1.

- **Standardization**  
  Mengubah nilai fitur sehingga memiliki rata-rata 0 dan varians 1.  
  Metode ini lebih tahan terhadap *outlier* dibandingkan normalisasi.


## 4. Pemilihan dan Pelatihan Model

Pada tahap ini, tujuan utama adalah menemukan model yang memberikan performa terbaik. Hindari langsung terpaku pada satu algoritma tertentu. Gunakan strategi *short-listing* untuk membandingkan beberapa model.

### 4.1 Mencoba Berbagai Model

Mulailah dengan mencoba beberapa algoritma Machine Learning, dari yang sederhana hingga yang lebih kompleks, antara lain:

- **Model Sederhana**
  - Linear Regression
  - Logistic Regression

- **Model Kompleks**
  - Random Forest
  - Support Vector Machine (SVM)
  - Gradient Boosting

Pendekatan ini membantu dalam memahami trade-off antara kompleksitas model dan performa.

### 4.2 Cross-Validation

Jangan hanya mengandalkan satu kali pembagian data (*train-test split*). Gunakan teknik **K-Fold Cross-Validation** untuk mendapatkan evaluasi performa yang lebih stabil dan representatif.

- Data dibagi menjadi *K* bagian (*fold*).
- Model dilatih sebanyak *K* kali.
- Setiap *fold* digunakan satu kali sebagai data validasi.
- Performa akhir diperoleh dari rata-rata hasil seluruh *fold*.

Cross-validation membantu mengurangi risiko *overfitting* dan memberikan gambaran performa model pada data yang berbeda-beda.


## 5. Fine-Tuning Model

Setelah mendapatkan kandidat model terbaik, langkah selanjutnya adalah melakukan *fine-tuning* untuk menemukan kombinasi **hyperparameter** yang paling optimal sehingga performa model dapat dimaksimalkan.

### 5.1 Grid Search

**Grid Search** dilakukan dengan mencoba seluruh kombinasi hyperparameter yang telah ditentukan secara manual.

- Menghasilkan hasil yang optimal jika ruang pencarian kecil.
- Membutuhkan waktu komputasi yang besar jika jumlah parameter banyak.
- Umumnya digunakan bersama *cross-validation*.

### 5.2 Randomized Search

**Randomized Search** memilih nilai hyperparameter secara acak dari ruang pencarian yang ditentukan.

- Lebih efisien dibandingkan Grid Search.
- Sangat efektif jika *search space* sangat luas.
- Cocok untuk model dengan banyak hyperparameter.

### 5.3 Ensemble Methods

**Ensemble Methods** menggabungkan beberapa model yang sudah memiliki performa baik untuk menghasilkan prediksi yang lebih akurat dan stabil.

- Mengurangi risiko *overfitting*.
- Meningkatkan generalisasi model.
- Contoh metode ensemble:
  - **Random Forest** (kumpulan dari banyak Decision Tree)
  - **Gradient Boosting**
  - **AdaBoost**


# 6. Contoh Implementasi: Pipeline Sederhana
Penggunaan Pipeline di Scikit-Learn sangat membantu agar proses transformasi data dan pelatihan model menjadi satu kesatuan yang rapi.

```python
import numpy as np
import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error

# --- SIMULASI DATA (Agar tidak NameError) ---
# Membuat data dummy seolah-olah data perumahan
data = {
    'income': [5.1, 3.2, 4.1, 7.5, 2.8, 8.1, 1.5, 6.2],
    'age': [32, 15, 41, 52, 22, 45, 28, 37],
    'rooms': [5, 4, np.nan, 6, 3, 7, 2, 5] # Ada data kosong (NaN)
}
housing = pd.DataFrame(data)
housing_labels = np.array([250000, 150000, 180000, 350000, 120000, 400000, 90000, 300000])

# --- MEMBANGUN PIPELINE ---
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy="median")), # Mengisi NaN dengan median
    ('std_scaler', StandardScaler()),              # Skalakan data (Mean 0, Std 1)
    ('forest_reg', RandomForestRegressor())        # Model utama
])

# --- PROSES TRAINING & OUTPUT ---
# 1. Melatih model
num_pipeline.fit(housing, housing_labels)

# 2. Melakukan Prediksi pada data yang sama (untuk contoh)
housing_predictions = num_pipeline.predict(housing)

# 3. Menghitung Error (RMSE)
mse = mean_squared_error(housing_labels, housing_predictions)
rmse = np.sqrt(mse)

print(f"Hasil Prediksi: {housing_predictions[:3]}")
print(f"Root Mean Squared Error (RMSE): {rmse}")
```

## 7. Evaluasi Akhir pada Test Set

Setelah model selesai dilatih dan menunjukkan performa yang baik pada data training, langkah selanjutnya adalah melakukan evaluasi akhir menggunakan **Test Set**. Tahap ini bertujuan untuk memastikan bahwa model tidak mengalami *overfitting* dan mampu melakukan generalisasi terhadap data yang belum pernah dilihat sebelumnya.

### 7.1 Generalization Error

- **Generalization Error** merupakan ukuran seberapa baik model bekerja pada data baru.
- Nilai error yang diperoleh dari **Test Set** (misalnya RMSE) menjadi estimasi performa model saat digunakan oleh pengguna nyata.
- Semakin kecil nilai error pada Test Set, semakin baik kemampuan generalisasi model.

### 7.2 Catatan Penting

- Jika performa model pada Test Set **jauh lebih buruk** dibandingkan data training, maka model mengalami **overfitting**.
- Untuk mengatasi hal tersebut, beberapa langkah yang dapat dilakukan:
  - Kembali ke tahap **Fine-Tuning Model**.
  - Menyederhanakan model.
  - Menambah data training.
  - Mengurangi kompleksitas fitur.

Evaluasi pada Test Set merupakan tahap terakhir sebelum model siap untuk diluncurkan ke lingkungan produksi.


# Chapter 3: Classification

## 1. Dataset MNIST

Bab ini berfokus pada **dataset MNIST**, yaitu kumpulan **70.000 gambar angka tulisan tangan (0–9)** yang dibuat oleh siswa SMA dan karyawan **Biro Sensus Amerika Serikat**. Dataset ini merupakan salah satu dataset paling populer untuk pembelajaran dan eksperimen di bidang *Machine Learning* dan *Computer Vision*.

### Karakteristik Dataset

- **Data**  
  Setiap gambar berukuran **28 × 28 piksel**, sehingga setiap sampel direpresentasikan sebagai **784 fitur numerik**.

- **Target**  
  Label berupa angka yang sesuai dengan gambar, yaitu **0 hingga 9**.

- **Tujuan**  
  Membangun model klasifikasi yang mampu **mengenali angka tulisan tangan** dengan tingkat akurasi yang tinggi.

Dataset MNIST sering digunakan sebagai *benchmark* untuk menguji dan membandingkan performa berbagai algoritma klasifikasi.


## 2. Klasifikasi Biner (Binary Classifier)

Sebelum menebak semua angka (0–9), kita mulai dengan masalah yang lebih sederhana: **“5-detector”**. Tujuannya adalah membedakan apakah sebuah gambar **angka 5** atau **bukan angka 5**.

### Output

- Hanya ada **dua kemungkinan**:  
  - `True`: gambar merupakan angka 5  
  - `False`: gambar bukan angka 5

### Algoritma yang Sering Digunakan

- **Stochastic Gradient Descent (SGD)**  
  - Efisien untuk menangani dataset besar seperti MNIST  
  - Sering digunakan sebagai titik awal sebelum mencoba model yang lebih kompleks


## 3. Metrik Performa (Lebih Kompleks dari Regresi)

Dalam klasifikasi, **akurasi** sering kali menipu, terutama jika dataset tidak seimbang. Oleh karena itu, kita membutuhkan metrik yang lebih detail untuk mengevaluasi performa model.

### 3.1 Confusion Matrix

Confusion Matrix menghitung berapa kali **kelas A diklasifikasikan sebagai kelas B**.

- **True Positive (TP)**: Benar menebak angka 5  
- **True Negative (TN)**: Benar menebak bukan angka 5  
- **False Positive (FP)**: Salah menebak angka lain sebagai angka 5 (*Type I Error*)  
- **False Negative (FN)**: Salah menebak angka 5 sebagai bukan angka 5 (*Type II Error*)

### 3.2 Precision dan Recall

- **Precision**: Akurasi dari prediksi positif  
  > “Saat model menebak angka 5, seberapa sering tebakan itu benar?”

- **Recall (Sensitivity)**: Rasio deteksi positif  
  > “Dari semua angka 5 yang ada, berapa banyak yang berhasil ditemukan model?”

### 3.3 F1 Score

- Kombinasi (harmonic mean) dari **Precision** dan **Recall**  
- Skor tinggi hanya jika **Precision** dan **Recall** keduanya tinggi


## 4. Precision/Recall Trade-off

Dalam klasifikasi, **meningkatkan Precision biasanya akan menurunkan Recall**, dan sebaliknya. Hal ini dikontrol oleh **threshold** keputusan klasifikasi.

### Threshold

- Keputusan model untuk mengklasifikasikan suatu sampel sebagai positif bergantung pada **skor probabilitas** dibandingkan ambang batas (*threshold*).  

### Dampak Threshold

- **Threshold dinaikkan** → Precision naik, Recall turun  
- **Threshold diturunkan** → Recall naik, Precision turun

Trade-off ini penting untuk disesuaikan dengan kebutuhan bisnis atau tujuan proyek.  
Contohnya:  
- Jika salah menolak angka 5 lebih kritis → fokus pada Recall tinggi  
- Jika salah menebak angka 5 lebih berisiko → fokus pada Precision tinggi


## 5. Klasifikasi Multikelas (Multiclass Classification)

Jika **Klasifikasi Biner** hanya membedakan dua kelas, **Multikelas** dapat membedakan lebih dari dua kelas, misalnya angka 0 hingga 9 pada MNIST.

### Strategi Multikelas

- **One-versus-All (OvA)**  
  - Membuat 10 klasifikasi biner, masing-masing mendeteksi satu angka  
  - Sistem memilih prediksi dengan **skor tertinggi**  
  - Paling umum digunakan karena efisien untuk dataset besar

- **One-versus-One (OvO)**  
  - Membuat klasifikasi untuk setiap pasangan angka (0 vs 1, 0 vs 2, dst.)  
  - Jumlah model meningkat drastis  
  - Efektif untuk algoritma yang sulit menangani dataset besar


## 6. Analisis Kesalahan (Error Analysis)

Untuk meningkatkan performa model, **tidak cukup hanya melihat akurasi**; kita perlu menganalisis di mana model melakukan kesalahan.

### Cara Analisis

- Menggunakan **Confusion Matrix** dalam bentuk **heatmap**  
- Memeriksa pola kesalahan, misalnya:  
  - Model sering tertukar antara angka **3 dan 5**  
  - Model sering tertukar antara angka **8 dan 9**

### Solusi Potensial

- Menambah data latih untuk angka yang sering salah diklasifikasikan  
- Melakukan **preprocessing gambar**, seperti:  
  - Menggeser (*shift*)  
  - Memutar (*rotation*)  
  - Normalisasi piksel

### **Contoh kode yaitu Melatih Classifier Biner**

```python
from sklearn.datasets import fetch_openml
import numpy as np

# 1. Load data MNIST
mnist = fetch_openml('mnist_784', version=1, as_frame=False)
X, y = mnist["data"], mnist["target"]

# 2. Ubah label dari string ke integer
y = y.astype(np.uint8)

# 3. Bagi data menjadi Training set (60rb) dan Test set (10rb)
X_train, X_test, y_train, y_test = X[:60000], X[60000:], y[:60000], y[60000:]
```

```python
from sklearn.linear_model import SGDClassifier

# Membuat label biner: True untuk angka 5, False untuk selain 5
y_train_5 = (y_train == 5)
y_test_5 = (y_test == 5)

# Inisialisasi dan latih model
sgd_clf = SGDClassifier(random_state=42)
sgd_clf.fit(X_train, y_train_5)

# Uji prediksi sederhana
some_digit = X[0] # Contoh gambar angka 5
print(f"Prediksi angka 5: {sgd_clf.predict([some_digit])}")
```

```python
from sklearn.linear_model import SGDClassifier
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import confusion_matrix, f1_score

# 1. Inisialisasi model SGD
sgd_clf = SGDClassifier(random_state=42)

# 2. Melatih model (X_train adalah gambar, y_train_5 adalah label biner)
sgd_clf.fit(X_train, y_train_5)

# 3. Prediksi dan Evaluasi dengan Confusion Matrix
y_train_pred = cross_val_predict(sgd_clf, X_train, y_train_5, cv=3)
matrix = confusion_matrix(y_train_5, y_train_pred)

print(f"Confusion Matrix:\n{matrix}")
print(f"F1 Score: {f1_score(y_train_5, y_train_pred)}")
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

# Visualisasi kesalahan dalam bentuk Heatmap
ConfusionMatrixDisplay.from_predictions(y_train_5, y_train_pred)
plt.show()
```

## 7. Metrik ROC dan AUC

Selain Precision dan Recall, metrik **ROC (Receiver Operating Characteristic)** sangat berguna untuk membandingkan performa berbagai model klasifikasi biner.

### ROC Curve

- Grafik yang memplot **True Positive Rate (TPR)** vs **False Positive Rate (FPR)**  
- Memberikan gambaran kemampuan model membedakan antara kelas positif dan negatif pada berbagai threshold.

### AUC (Area Under the Curve)

- Mengukur luas di bawah ROC Curve  
- Interpretasi:  
  - **AUC = 1** → model sempurna  
  - **AUC = 0.5** → model menebak secara acak

### Kapan Menggunakan PR vs ROC?

- **Precision-Recall (PR) Curve** → gunakan jika:  
  - Kelas positif jarang (*imbalanced dataset*)  
  - Fokus pada mengurangi **False Positives**

- **ROC Curve** → gunakan untuk kasus umum atau dataset seimbang

## 8. Kesimpulan Klasifikasi

Inti dari Chapter 3:  

- **Pilih metrik yang sesuai** (Precision, Recall, F1 Score) berdasarkan kebutuhan bisnis.  
- **Gunakan Error Analysis** (misalnya Confusion Matrix heatmap) untuk memperbaiki performa secara spesifik pada kelas yang sering tertukar.  
- **Pertimbangkan trade-off**:  
  - Apakah ingin model yang **sangat selektif** (Precision tinggi)  
  - Atau model yang **sangat sensitif** (Recall tinggi)

Klasifikasi bukan hanya soal akurasi tinggi, tetapi juga soal memilih strategi dan metrik yang tepat untuk masalah dan data yang dihadapi.
