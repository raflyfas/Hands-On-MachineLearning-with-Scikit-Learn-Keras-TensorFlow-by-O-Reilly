## 1. From Biological to Artificial Neurons

- Neuron biologis menerima sinyal melalui dendrit, mengintegrasikan sinyal, dan menembakkan impuls jika melampaui ambang tertentu.[file:1]
- McCulloch & Pitts memodelkan **artificial neuron** biner dengan input dan output 0/1 yang bisa menyusun operasi logika seperti AND, OR, NOT, XOR (dengan kombinasi beberapa neuron).[file:1]
- Perceptron (Rosenblatt, 1957) memperkenalkan **Threshold Logic Unit (TLU)** yang menghitung jumlah berbobot \(z = x \cdot w\) lalu menerapkan step function untuk menghasilkan output 0/1.[file:1]
- Keterbatasan Perceptron: hanya bisa memisahkan data yang **linearly separable**, sehingga gagal di problem sederhana seperti XOR; masalah ini diatasi dengan **Multilayer Perceptron (MLP)** yang menambahkan hidden layer non‑linear.[file:1]


## 2. Logical Computations with Neurons
McCulloch dan Pitts memodelkan neuron buatan dengan beberapa input biner (on/off) dan satu output biner, yang aktif jika jumlah input aktif melewati ambang tertentu. Dengan neuron sederhana seperti ini, mereka menunjukkan bahwa jaringan neuron buatan dapat merepresentasikan operasi logika seperti identity, AND, OR, dan kombinasi yang lebih kompleks (misalnya A aktif dan B tidak aktif, yang setara dengan A ∧ ¬B).
​

Jaringan–jaringan logika ini bisa dikombinasikan untuk membangun ekspresi logika apapun, sehingga ANN pada dasarnya mampu melakukan komputasi logis umum.
​

## 3. The Perceptron

Perceptron adalah arsitektur **Artificial Neural Network (ANN)** sederhana yang menggunakan neuron bernama **Threshold Logic Unit (TLU)** atau **Linear Threshold Unit (LTU)**.

Setiap TLU menghitung jumlah berbobot dari input sebagai berikut:

$$
z = x^\top w
$$

Kemudian, nilai tersebut diproses menggunakan fungsi langkah (*step function*), seperti **Heaviside** atau **sign**, untuk menghasilkan output biner:

$$
h(x) = step(z)
$$

Perceptron terdiri dari **satu lapisan TLU** yang terhubung penuh (*fully connected*) ke neuron input, ditambah **satu neuron bias** yang selalu bernilai 1. Oleh karena itu, perceptron cocok digunakan sebagai **model klasifikasi linear**, baik untuk kasus **biner** maupun **multi-output**.

### Learning Rule

Algoritma pembelajaran perceptron memperbarui bobot menggunakan aturan **Hebbian termodifikasi**. Koneksi yang membantu mengurangi kesalahan (*error*) akan diperkuat berdasarkan aturan berikut:

$$
w_{i,j} \leftarrow w_{i,j} + \eta (y_j - \hat{y}_j) x_i
$$

dengan:
- $w_{i,j}$ : bobot dari neuron input ke-$i$ ke neuron output ke-$j$
- $\eta$ : *learning rate*
- $y_j$ : nilai target
- $\hat{y}_j$ : nilai prediksi
- $x_i$ : nilai input ke-$i$


## Example: Perceptron with Scikit-Learn

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.linear_model import Perceptron

iris = load_iris()
X = iris.data[:, (2, 3)]  # petal length, petal width
y = (iris.target == 0).astype(np.int32)  # Iris setosa?

per_clf = Perceptron()
per_clf.fit(X, y)

y_pred = per_clf.predict([[2, 0.5]])
print(y_pred)
```

Perceptron hanya bisa memisahkan data yang linearly separable, dan tidak bisa menyelesaikan kasus XOR, sehingga digantikan oleh arsitektur multilayer yang lebih kuat

## 4. The Multilayer Perceptron and Backpropagation

**Multilayer Perceptron (MLP)** adalah pengembangan dari perceptron tunggal yang terdiri dari:
- satu **input layer** (bersifat *passthrough*),
- satu atau lebih **hidden layer**, dan
- satu **output layer**.

Seluruh neuron antar lapisan terhubung penuh (*fully connected*), dengan **neuron bias** pada setiap lapisan kecuali output layer. Aliran sinyal hanya bergerak satu arah, dari input menuju output, sehingga MLP dikategorikan sebagai **feedforward neural network**.  
Jika jumlah *hidden layer* cukup banyak, jaringan ini disebut **Deep Neural Network (DNN)**.



### **Backpropagation** Algorithm

Algoritma pelatihan utama pada MLP adalah **backpropagation**, yaitu metode **Gradient Descent** yang dikombinasikan dengan perhitungan gradien otomatis menggunakan **reverse-mode automatic differentiation**.

Proses pelatihan terdiri dari dua tahap utama:

1. **Forward Pass**  
   Input dilewatkan melalui jaringan untuk menghasilkan output, kemudian dihitung nilai *loss* berdasarkan selisih antara prediksi dan target.

2. **Backward Pass**  
   Gradien kesalahan (*error*) terhadap setiap bobot dan bias dihitung menggunakan **aturan rantai (chain rule)**. Gradien ini kemudian digunakan untuk memperbarui parameter jaringan melalui langkah **Gradient Descent**.

### Activation Functions

Berbeda dengan perceptron tunggal yang menggunakan fungsi langkah (*step function*), MLP menggunakan **fungsi aktivasi yang terdiferensiasi** agar gradien dapat dihitung dengan baik dan proses pelatihan berjalan stabil.

Fungsi aktivasi yang umum digunakan antara lain:
- **Sigmoid**
- **Hyperbolic Tangent (tanh)**
- **Rectified Linear Unit (ReLU)**

Penggunaan fungsi-fungsi ini memungkinkan jaringan mempelajari hubungan non-linear yang kompleks pada data.


## 5. Activation Functions

Fungsi aktivasi berperan memberikan **non-linearitas** pada jaringan saraf. Tanpa non-linearitas, komposisi beberapa lapisan hanya akan runtuh menjadi **satu transformasi linear**, sehingga bahkan jaringan yang sangat dalam tetap setara dengan satu layer linear dan tidak mampu mempelajari pola yang kompleks.

Dengan penggunaan fungsi aktivasi **non-linear**, *Deep Neural Network (DNN)* secara teoretis mampu mengaproksimasi **fungsi kontinu apa pun** (*universal approximation*).

---

### Sigmoid

Fungsi **sigmoid** didefinisikan sebagai:

$$
\sigma(z) = \frac{1}{1 + \exp(-z)}
$$

Sigmoid menghasilkan output pada rentang **(0, 1)**. Fungsi ini memiliki peran historis yang penting, terutama pada model awal neural network, namun cenderung mengalami **saturasi gradien** pada nilai ekstrem sehingga kurang cocok untuk jaringan dalam.

---

### Hyperbolic Tangent (tanh)

Fungsi **tanh** didefinisikan sebagai:

$$
\tanh(z) = 2\sigma(2z) - 1
$$

Rentang output tanh adalah **[-1, 1]**, sehingga nilai aktivasi awal lebih terpusat di sekitar nol. Hal ini sering membantu **mempercepat konvergensi** dibandingkan sigmoid.

---

### Rectified Linear Unit (ReLU)

Fungsi **ReLU** didefinisikan sebagai:

$$
\text{ReLU}(z) = \max(0, z)
$$

ReLU bersifat **sederhana dan cepat dihitung**, serta menjadi pilihan default pada *hidden layer* modern. Namun, fungsi ini tidak terdiferensiasi di titik 0 dan memiliki **gradien nol untuk $z < 0$**, yang dapat menyebabkan fenomena *dead neurons*.


## 6. Regression MLPs

Untuk tugas **regresi**, arsitektur **Multilayer Perceptron (MLP)** umumnya disusun sebagai berikut.

### Architecture

- **Input layer**  
  Satu neuron untuk setiap fitur input.

- **Hidden layers**  
  Beberapa *hidden layer* (biasanya 1–5 lapisan) dengan sekitar **10–100 neuron per layer**, menggunakan fungsi aktivasi **ReLU** atau **SELU**.

- **Output layer**  
  Satu neuron untuk setiap dimensi target, dengan pilihan aktivasi sebagai berikut:
  - **Tanpa aktivasi (linear)** untuk regresi umum
  - **ReLU** atau **softplus** jika output harus bernilai positif
  - **Sigmoid** atau **tanh** jika output dibatasi (*bounded*), dengan label target diskalakan ke rentang yang sesuai

---

### Loss Functions

Fungsi *loss* yang umum digunakan pada regresi meliputi:
- **Mean Squared Error (MSE)** sebagai pilihan standar
- **Mean Absolute Error (MAE)** atau **Huber Loss** jika data mengandung banyak *outlier*

Pemilihan fungsi *loss* sangat memengaruhi sensitivitas model terhadap kesalahan ekstrem dan stabilitas proses pelatihan.


```python
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow import keras

housing = fetch_california_housing()

X_train_full, X_test, y_train_full, y_test = train_test_split(
    housing.data, housing.target, random_state=42
)
X_train, X_valid, y_train, y_valid = train_test_split(
    X_train_full, y_train_full, random_state=42
)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_valid = scaler.transform(X_valid)
X_test = scaler.transform(X_test)

model = keras.models.Sequential([
    keras.layers.Dense(30, activation="relu", input_shape=X_train.shape[1:]),
    keras.layers.Dense(1)
])

model.compile(loss="mean_squared_error", optimizer="sgd")
history = model.fit(X_train, y_train, epochs=20,
                    validation_data=(X_valid, y_valid))
mse_test = model.evaluate(X_test, y_test)
X_new = X_test[:3]
y_pred = model.predict(X_new)
```

## 7. Classification MLPs

Untuk **klasifikasi biner**, MLP biasanya menggunakan **satu neuron output** dengan fungsi aktivasi **sigmoid**, sehingga output dapat diinterpretasikan sebagai **probabilitas kelas positif**.

Pada **klasifikasi multilabel biner** (beberapa label yang saling independen), digunakan **satu neuron sigmoid per label**. Dalam kasus ini, probabilitas antar label **tidak harus berjumlah 1**, karena setiap label dapat aktif secara independen.

Untuk **klasifikasi multiclass eksklusif** (misalnya 10 kelas pada dataset MNIST atau Fashion MNIST), digunakan:
- **Satu neuron output per kelas**
- Fungsi aktivasi **softmax**, sehingga probabilitas setiap kelas berada pada rentang (0, 1) dan **total probabilitas berjumlah 1**

Fungsi *loss* yang umum digunakan adalah **cross-entropy (log loss)**, karena model memprediksi **distribusi probabilitas kelas**.


## 8. Implementing MLPs with Keras (Fashion MNIST, Sequential API)

Subbab ini menjelaskan implementasi **Multilayer Perceptron (MLP)** menggunakan **Keras (tf.keras)** untuk tugas klasifikasi pada dataset **Fashion MNIST**.

Dataset Fashion MNIST terdiri dari **70.000 gambar grayscale berukuran 28×28 piksel** yang terbagi ke dalam **10 kelas** item fashion. Data biasanya dipisahkan menjadi:
- **Training set**
- **Validation set** (dibuat secara manual dari `train_full`)
- **Test set**

Sebelum pelatihan, nilai piksel diskalakan agar berada pada rentang yang sesuai untuk proses optimisasi.

---

### Model Architecture

Model dibangun menggunakan **Sequential API**, dengan susunan sebagai berikut:
- **Flatten layer** untuk mengubah gambar 28×28 menjadi vektor berdimensi 784
- Dua **Dense hidden layer** dengan fungsi aktivasi **ReLU**
- **Output layer** berupa Dense dengan **10 neuron** dan aktivasi **softmax**

---

### Training and Evaluation

Model dikompilasi menggunakan:
- **Loss function**: `sparse_categorical_crossentropy`
- **Optimizer**: Stochastic Gradient Descent (SGD)
- **Metric**: accuracy

Pelatihan dilakukan menggunakan metode `fit()`, sementara evaluasi kinerja model dilakukan dengan `evaluate()`. Prediksi kelas dapat diperoleh menggunakan metode `predict()`.

### **Example: Fashion MNIST Classification with Sequential API**

```python
import tensorflow as tf
from tensorflow import keras
import numpy as np

# Load dataset
fashion_mnist = keras.datasets.fashion_mnist
(X_train_full, y_train_full), (X_test, y_test) = fashion_mnist.load_data()

# Scale to [0, 1] and create validation set
X_train_full = X_train_full / 255.0
X_test = X_test / 255.0

X_valid, X_train = X_train_full[:5000], X_train_full[5000:]
y_valid, y_train = y_train_full[:5000], y_train_full[5000:]

class_names = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
               "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]

# Build model
model = keras.models.Sequential([
    keras.layers.Flatten(input_shape=[28, 28]),
    keras.layers.Dense(300, activation="relu"),
    keras.layers.Dense(100, activation="relu"),
    keras.layers.Dense(10, activation="softmax")
])

# Compile model
model.compile(loss="sparse_categorical_crossentropy",
              optimizer="sgd",
              metrics=["accuracy"])

# Train model
history = model.fit(X_train, y_train, epochs=30,
                    validation_data=(X_valid, y_valid))

# Evaluate on test set
test_loss, test_acc = model.evaluate(X_test, y_test)

# Predict some examples
X_new = X_test[:3]
y_proba = model.predict(X_new)
y_pred = np.argmax(y_proba, axis=1)

print("Predicted classes:", y_pred)
print("True classes     :", y_test[:3])
```

## 9. Building Complex Models Using the Functional API
Functional API digunakan saat arsitektur lebih kompleks dari sekadar stack berurutan, misalnya Wide & Deep network, multiple inputs, atau multiple outputs. Wide & Deep menggabungkan jalur “deep” (beberapa hidden layer) dengan shortcut dari input langsung ke output, sehingga jaringan bisa belajar pola kompleks dan aturan sederhana sekaligus.
​

Dengan banyak input, sebagian fitur bisa dikirim ke jalur “wide” dan subset lain ke jalur “deep” lalu dikonkatenasi, sedangkan dengan banyak output, satu model bisa menyelesaikan beberapa tugas sekaligus (misalnya regresi dan klasifikasi), atau menambah auxiliary output sebagai regularizer.
​

## **Example: Wide & Deep Regression (Multiple Inputs)**

```python
from tensorflow import keras

input_A = keras.layers.Input(shape=[5], name="wide_input")
input_B = keras.layers.Input(shape=[6], name="deep_input")

hidden1 = keras.layers.Dense(30, activation="relu")(input_B)
hidden2 = keras.layers.Dense(30, activation="relu")(hidden1)

concat = keras.layers.concatenate([input_A, hidden2])
output = keras.layers.Dense(1, name="output")(concat)

model = keras.Model(inputs=[input_A, input_B], outputs=[output])

model.compile(loss="mse", optimizer=keras.optimizers.SGD(learning_rate=1e-3))

# Example split of features:
# X_train_A = X_train[:, :5]
# X_train_B = X_train[:, 2:]
# history = model.fit((X_train_A, X_train_B), y_train, epochs=20,
#                     validation_data=((X_valid_A, X_valid_B), y_valid))
```

## 10. Saving Models, Callbacks, TensorBoard, and Hyperparameter Tuning

Keras menyediakan mekanisme yang mudah untuk **menyimpan dan memulihkan model**, baik dalam bentuk **satu file lengkap** maupun **hanya bobot model**.

Model dapat disimpan dan dimuat kembali menggunakan:
- `model.save()` dan `load_model()` untuk menyimpan seluruh arsitektur, bobot, dan konfigurasi training
- `save_weights()` dan `load_weights()` untuk menyimpan dan memuat bobot saja

Pendekatan ini memudahkan *deployment*, eksperimen ulang, serta kelanjutan pelatihan tanpa harus melatih model dari awal.

---

### Callbacks

Untuk proses pelatihan yang memakan waktu lama, Keras menyediakan **callbacks** yang membantu mengontrol dan memantau training secara otomatis. Callback yang umum digunakan antara lain:
- **ModelCheckpoint**, untuk menyimpan *checkpoint* model secara otomatis selama training
- **EarlyStopping**, untuk menghentikan training ketika performa pada data validasi tidak lagi membaik

Penggunaan callbacks membantu mencegah *overfitting* dan menghemat sumber daya komputasi.

---

### TensorBoard

**TensorBoard** menyediakan alat visualisasi untuk memantau dan menganalisis proses pelatihan model, seperti:
- *Learning curves* (loss dan accuracy)
- Grafik arsitektur model
- Histogram bobot dan aktivasi
- Metode optimisasi dan distribusi gradien

TensorBoard dapat digunakan melalui **TensorBoard callback** atau API **`tf.summary`**.

---

### Hyperparameter Tuning

Untuk meningkatkan performa model, sering kali diperlukan **tuning hiperparameter**, seperti:
- Jumlah layer dan neuron
- Learning rate
- Batch size
- Optimizer
- Fungsi aktivasi

Proses tuning dapat dilakukan menggunakan:
- **RandomizedSearchCV** dengan *wrapper* `KerasRegressor` atau `KerasClassifier`
- Library khusus seperti **Hyperopt**, **Keras Tuner**, dan alat serupa lainnya

Pendekatan ini memungkinkan eksplorasi ruang hiperparameter secara sistematis dan efisien.


## **Example: Saving, EarlyStopping, and Checkpoints**

```python
from tensorflow import keras

# Save full model
model.save("my_keras_model.h5")

# Load model
model = keras.models.load_model("my_keras_model.h5")

# Checkpoint callback (save best model on validation)
checkpoint_cb = keras.callbacks.ModelCheckpoint(
    "best_model.h5", save_best_only=True
)

# Early stopping callback
early_stopping_cb = keras.callbacks.EarlyStopping(
    patience=10, restore_best_weights=True
)

history = model.fit(
    X_train, y_train,
    epochs=100,
    validation_data=(X_valid, y_valid),
    callbacks=[checkpoint_cb, early_stopping_cb]
)
```

## **Example: Basic TensorBoard Logging**

```python
import os
import time
import numpy as np
import tensorflow as tf
from tensorflow import keras

def get_run_logdir(root_logdir="my_logs"):
    run_id = time.strftime("run_%Y_%m_%d-%H_%M_%S")
    return os.path.join(root_logdir, run_id)

log_dir = get_run_logdir()
tensorboard_cb = keras.callbacks.TensorBoard(log_dir)

history = model.fit(
    X_train, y_train,
    epochs=20,
    validation_data=(X_valid, y_valid),
    callbacks=[tensorboard_cb]
)

# Lower-level summary example
writer = tf.summary.create_file_writer(log_dir)
with writer.as_default():
    for step in range(1, 1001):
        tf.summary.scalar("my_scalar", np.sin(step / 10), step=step)
```
