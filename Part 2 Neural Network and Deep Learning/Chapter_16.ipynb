## 1. Character RNN and Shakespeare Text Generation

Bab ini diawali dengan **Character-level RNN (Char-RNN)**, yaitu model RNN yang dilatih untuk memprediksi **karakter berikutnya** dalam teks Shakespeare. Dengan mempelajari distribusi karakter, model mampu menghasilkan teks baru secara autoregresif, satu karakter demi satu.

Karena korpus teks sangat panjang (lebih dari satu juta karakter), data tidak diproses sekaligus. Sebaliknya, teks dipecah menjadi banyak **window pendek** menggunakan `tf.data.Dataset.window()` dan **truncated Backpropagation Through Time (BPTT)**, misalnya urutan sepanjang 100 karakter sebagai input dan karakter ke-101 sebagai target.

Setelah window dibuat, dataset:
- diratakan dengan `flat_map`,
- di-*shuffle*,
- di-*batch*,
- lalu dipisahkan menjadi input (semua kecuali karakter terakhir) dan target (semua kecuali karakter pertama).

Setiap karakter direpresentasikan dengan **one-hot encoding** (`tf.one_hot`). Model yang digunakan adalah **GRU dua lapis** dengan `TimeDistributed(Dense)` dan aktivasi softmax, dilatih menggunakan loss `sparse_categorical_crossentropy` untuk memprediksi probabilitas seluruh karakter (misalnya 39 karakter) di setiap time step.

---

### Stateless vs Stateful RNN

Pada **stateless RNN**, hidden state selalu di-reset ke nol pada setiap batch. Akibatnya, model hanya dapat mempelajari ketergantungan sepanjang window yang di-*BPTT*.

Untuk mempelajari pola yang lebih panjang tanpa melakukan backpropagation pada seluruh teks, diperkenalkan **stateful RNN**. Pada pendekatan ini, hidden state akhir suatu batch digunakan sebagai state awal batch berikutnya.

Agar stateful RNN bekerja dengan benar:
- batch harus benar-benar berurutan,
- window dibuat **non-overlap** (shift = `n_steps`),
- batch size biasanya kecil (bahkan 1 untuk setup paling sederhana),
- layer RNN dibuat dengan `stateful=True` dan `batch_input_shape=[batch_size, None, vocab_size]`,
- state di-reset secara eksplisit di awal setiap epoch menggunakan callback khusus.

## 2. Stateful RNN for Longer Dependencies

Pada **stateless RNN**, hidden state selalu di-reset ke nol pada setiap batch. Konsekuensinya, model hanya mampu mempelajari ketergantungan yang terbatas pada panjang window yang di-*unroll* selama Backpropagation Through Time (BPTT).

Untuk menangkap **ketergantungan jangka lebih panjang** tanpa harus melakukan backpropagation pada seluruh urutan teks, diperkenalkan konsep **stateful RNN**. Pada pendekatan ini, hidden state akhir dari suatu batch dipertahankan dan digunakan kembali sebagai state awal untuk batch berikutnya, sehingga informasi dapat mengalir lintas batch.

Agar stateful RNN bekerja secara benar dan konsisten, beberapa syarat penting harus dipenuhi:
- Batch harus berisi **urutan yang benar-benar saling berlanjut**, bukan hasil shuffle acak.
- Dataset dibentuk menggunakan **window non-overlap**, dengan `shift = n_steps`.
- **Batch size kecil** sangat dianjurkan (bahkan 1 untuk konfigurasi paling sederhana).
- Layer RNN dikonfigurasi dengan `stateful=True` dan  
  `batch_input_shape = [batch_size, None, vocab_size]`.
- Hidden state harus **di-reset secara eksplisit** pada awal setiap epoch, biasanya menggunakan callback khusus.

Dengan pengaturan ini, stateful RNN mampu memodelkan struktur dan dependensi jangka panjang secara lebih efektif dibanding stateless RNN, tanpa biaya komputasi backpropagation yang berlebihan.


## **Example: Dataset dan Model GRU Stateful**

```python
batch_size = 1  # sederhana: satu sequence per batch
n_steps = 100
window_length = n_steps + 1

dataset = tf.data.Dataset.from_tensor_slices(encoded[:train_size])
dataset = dataset.window(window_length, shift=n_steps, drop_remainder=True)
dataset = dataset.flat_map(lambda window: window.batch(window_length))
dataset = dataset.batch(batch_size)
dataset = dataset.map(lambda windows: (windows[:, :-1], windows[:, 1:]))
dataset = dataset.map(
    lambda X_batch, Y_batch: (tf.one_hot(X_batch, depth=max_id), Y_batch)
)
dataset = dataset.prefetch(1)

model = keras.models.Sequential([
    keras.layers.GRU(
        128, return_sequences=True, stateful=True,
        dropout=0.2, recurrent_dropout=0.2,
        batch_input_shape=[batch_size, None, max_id]
    ),
    keras.layers.GRU(
        128, return_sequences=True, stateful=True,
        dropout=0.2, recurrent_dropout=0.2
    ),
    keras.layers.TimeDistributed(
        keras.layers.Dense(max_id, activation="softmax")
    )
])

class ResetStatesCallback(keras.callbacks.Callback):
    def on_epoch_begin(self, epoch, logs=None):
        self.model.reset_states()

model.compile(loss="sparse_categorical_crossentropy", optimizer="adam")
history = model.fit(dataset, epochs=50, callbacks=[ResetStatesCallback()])
```

## 3. Sentiment Analysis with Word-Level RNNs

Setelah membahas pemodelan pada tingkat karakter, bab ini beralih ke **word-level RNN** untuk tugas **sentiment analysis**, menggunakan dataset **IMDb** yang berisi sekitar 50.000 ulasan film.

Pada pendekatan ini, setiap kata direpresentasikan sebagai **ID numerik**, baik dengan memanfaatkan dataset IMDb bawaan Keras maupun dengan membangun **vocabulary kustom** menggunakan `Tokenizer` atau `tf.lookup.StaticVocabularyTable`. ID kata tersebut kemudian dipetakan ke vektor kontinu berdimensi rendah melalui **Embedding layer**, sehingga model dapat mempelajari representasi semantik kata secara end-to-end.

Untuk membangun vocabulary kustom, seluruh teks diproses satu kali untuk menghitung frekuensi kata (misalnya dengan `Counter`). Selanjutnya dipilih sejumlah kata paling sering muncul (misalnya 10.000 kata), sementara kata lain dipetakan ke **bucket out-of-vocabulary (OOV)**.

Pipeline pemrosesan data secara keseluruhan adalah sebagai berikut:
- dataset di-*batch*,
- token teks di-*preprocess*,
- token di-*lookup* menjadi ID,
- urutan ID dimasukkan ke **model RNN dua lapis berbasis GRU**,
- satu neuron output dengan aktivasi **sigmoid** digunakan untuk memprediksi sentimen positif atau negatif.

Pendekatan word-level ini memungkinkan model menangkap konteks semantik yang lebih kaya dibanding character-level modeling, sehingga umumnya memberikan performa yang lebih baik pada tugas analisis sentimen.


## **Example: Build Vocabulary dan Model Sentiment Analysis**

```python
import tensorflow_datasets as tfds
from collections import Counter
from tensorflow import keras
import tensorflow as tf

datasets, info = tfds.load("imdb_reviews", as_supervised=True, with_info=True)

def preprocess(X_batch, y_batch):
    # lowercase & split jadi tokens bytes
    X_batch = tf.strings.regex_replace(X_batch, rb"<br\s*/?>", b" ")
    X_batch = tf.strings.regex_replace(X_batch, rb"[^a-zA-Z']", b" ")
    X_batch = tf.strings.lower(X_batch)
    return tf.strings.split(X_batch), y_batch

vocabulary = Counter()
for X_batch, y_batch in datasets["train"].batch(32).map(preprocess):
    for review in X_batch:
        vocabulary.update(list(review.numpy()))

vocab_size = 10000
truncated_vocabulary = [
    word for word, count in vocabulary.most_common()[:vocab_size]
]

words = tf.constant(truncated_vocabulary)
word_ids = tf.range(len(truncated_vocabulary), dtype=tf.int64)
vocab_init = tf.lookup.KeyValueTensorInitializer(words, word_ids)
num_oov_buckets = 1000
table = tf.lookup.StaticVocabularyTable(vocab_init, num_oov_buckets)

def encode_words(X_batch, y_batch):
    return table.lookup(X_batch), y_batch

train_set = datasets["train"].batch(32).map(preprocess)
train_set = train_set.map(encode_words).prefetch(1)

embed_size = 128
model = keras.models.Sequential([
    keras.layers.Embedding(vocab_size + num_oov_buckets, embed_size,
                           input_shape=[None]),
    keras.layers.GRU(128, return_sequences=True),
    keras.layers.GRU(128),
    keras.layers.Dense(1, activation="sigmoid")
])
model.compile(loss="binary_crossentropy", optimizer="adam",
              metrics=["accuracy"])
history = model.fit(train_set, epochs=5)
```

## 4. Masking, Pretrained Embeddings, and Variable-Length Sequences

Ulasan teks memiliki panjang yang bervariasi, sehingga perlu dilakukan **padding** agar dapat diproses dalam batch. Untuk mencegah model belajar dari token padding (biasanya direpresentasikan dengan ID 0), digunakan mekanisme **masking**, sehingga token tersebut diabaikan saat perhitungan hidden state dan loss.

Pendekatan paling sederhana adalah dengan mengaktifkan opsi  
`mask_zero=True` pada **Embedding layer**. Dengan cara ini, Keras secara otomatis membuat mask untuk token bernilai nol dan meneruskannya ke layer berikutnya, seperti RNN atau `TimeDistributed`, yang mendukung masking.

Pada arsitektur yang lebih kompleks—misalnya kombinasi **Conv1D dan GRU**—masking sering perlu ditangani secara manual. Mask dapat dihitung menggunakan ekspresi seperti `K.not_equal(inputs, 0)` di dalam `Lambda` layer, kemudian diteruskan secara eksplisit ke layer GRU melalui **Functional API**.

Selain pendekatan berbasis RNN, bab ini juga memperkenalkan penggunaan **pretrained sentence embeddings** dari **TensorFlow Hub** sebagai alternatif yang lebih sederhana. Contohnya, model `nnlm-en-dim50` dapat digunakan melalui `hub.KerasLayer` untuk:
- menerima input berupa string mentah,
- langsung menghasilkan embedding kalimat berdimensi tetap,
- kemudian hanya memerlukan satu atau beberapa `Dense` layer untuk tugas sentiment analysis.

Pendekatan berbasis pretrained embeddings ini sering memberikan hasil yang kompetitif, terutama ketika data berlabel terbatas, serta secara signifikan menyederhanakan arsitektur model.


## **Example: Masking Manual dan Pretrained Sentence Embeddinng**

```python
from tensorflow import keras
import tensorflow.keras.backend as K
import tensorflow_hub as hub

# Masking manual di Functional API
vocab_size = 10000; num_oov_buckets = 1000; embed_size = 128
inputs = keras.layers.Input(shape=[None])
mask = keras.layers.Lambda(lambda x: K.not_equal(x, 0))(inputs)
z = keras.layers.Embedding(vocab_size + num_oov_buckets, embed_size)(inputs)
z = keras.layers.GRU(128, return_sequences=True)(z, mask=mask)
z = keras.layers.GRU(128)(z, mask=mask)
outputs = keras.layers.Dense(1, activation="sigmoid")(z)
model_mask = keras.Model(inputs=inputs, outputs=outputs)

# Model dengan sentence embedding dari TF Hub
hub_layer = hub.KerasLayer(
    "https://tfhub.dev/google/tf2-preview/nnlm-en-dim50/1",
    dtype=tf.string, input_shape=[], output_shape=[50]
)
model_hub = keras.Sequential([
    hub_layer,
    keras.layers.Dense(128, activation="relu"),
    keras.layers.Dense(1, activation="sigmoid")
])
model_hub.compile(loss="binary_crossentropy", optimizer="adam",
                  metrics=["accuracy"])
train_set = datasets["train"].batch(32).prefetch(1)
history = model_hub.fit(train_set, epochs=5)
```

## 5. Encoder–Decoder for Machine Translation and Beam Search

Untuk tugas **Neural Machine Translation (NMT)**, bab ini menggunakan arsitektur **encoder–decoder**. Encoder bertugas memproses kalimat sumber (misalnya bahasa Inggris) dan merangkum informasinya ke dalam representasi state, sedangkan decoder menghasilkan kalimat target (misalnya bahasa Prancis) secara autoregresif, satu kata pada setiap langkah waktu.

Selama proses training, digunakan teknik **teacher forcing**, di mana input decoder adalah kalimat target yang digeser satu langkah ke kanan. Proses ini biasanya diawali dengan token khusus **SOS (start-of-sequence)**. Model kemudian dilatih menggunakan loss `sparse_categorical_crossentropy` untuk memaksimalkan probabilitas distribusi kata target yang benar pada setiap langkah.

Untuk implementasi praktis, **TensorFlow Addons** menyediakan API `seq2seq` yang memudahkan pembangunan model encoder–decoder, termasuk komponen seperti:
- `LSTMCell` untuk unit rekuren,
- `TrainingSampler` untuk mengatur input decoder saat training,
- dan `BasicDecoder` sebagai pengendali proses decoding.

Pada tahap inference, strategi **greedy decoding**—memilih kata dengan probabilitas tertinggi di setiap langkah—sering kali menghasilkan terjemahan yang kurang optimal. Oleh karena itu, digunakan **beam search**, yaitu teknik decoding yang:
- mempertahankan beberapa kandidat terjemahan terbaik secara simultan,
- memperluas semua kandidat tersebut di setiap langkah,
- lalu menyimpan hanya `beam_width` urutan dengan probabilitas gabungan tertinggi.

Beam search secara konsisten menghasilkan terjemahan yang lebih akurat dan natural dibanding greedy decoding, terutama pada kalimat yang panjang dan kompleks.

## 6. Attention Mechanisms and Transformer Architecture

Mekanisme **attention** memungkinkan decoder untuk secara dinamis memfokuskan perhatian pada bagian input yang paling relevan di setiap langkah waktu. Dengan demikian, jalur informasi dari kata sumber ke kata target menjadi jauh lebih pendek, yang secara signifikan meningkatkan kualitas terjemahan—terutama untuk kalimat yang panjang.

Pada **Bahdanau attention** (additive atau concatenative), setiap output encoder dan hidden state decoder digunakan untuk menghitung skor kesesuaian  
e(t, i) melalui sebuah jaringan kecil. Skor ini kemudian dinormalisasi menggunakan softmax menjadi bobot perhatian α(t, i). Context vector dihitung sebagai kombinasi linear dari seluruh output encoder:

\[
\text{context}_t = \sum_i \alpha(t, i)\, y(i)
\]

Context vector ini kemudian digabungkan dengan hidden state decoder untuk menghasilkan prediksi kata berikutnya.

Sebagai alternatif yang lebih efisien, **Luong attention** (multiplicative) menghitung skor perhatian menggunakan **dot product** antara hidden state decoder dan output encoder, atau menggunakan versi *general* yang menambahkan transformasi linear. Pendekatan ini lebih ringan secara komputasi dan sering memberikan performa yang lebih baik. TensorFlow Addons menyediakan implementasi siap pakai seperti `LuongAttention` dan `AttentionWrapper` untuk melengkapi decoder dengan mekanisme attention.

---

### Transformer Architecture

Bab ini kemudian beralih ke **Transformer**, sebuah arsitektur yang sepenuhnya menghilangkan RNN dan CNN, dan hanya mengandalkan **attention** sebagai mekanisme utama pemodelan dependensi.

Transformer dibangun dari beberapa komponen inti:
- **Multi-Head Attention**
- **Position-wise Feed-Forward Networks**
- **Layer Normalization**
- **Skip (Residual) Connections**
- **Positional Encoding** berbasis fungsi sinus–kosinus untuk memasukkan informasi urutan.

Inti dari Transformer adalah **Scaled Dot-Product Attention**, yang dirumuskan sebagai:

\[
\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^\top}{\sqrt{d_k}}\right)V
\]

Pada **Multi-Head Attention**, query (Q), key (K), dan value (V) terlebih dahulu diproyeksikan ke beberapa ruang representasi berbeda melalui transformasi linear. Setiap *head* mempelajari pola perhatian yang berbeda—misalnya relasi posisi, struktur sintaks, atau aspek semantik—dan hasilnya kemudian digabungkan kembali.

Dengan kemampuan memodelkan dependensi jangka panjang secara paralel dan efisien, Transformer menjadi fondasi bagi hampir seluruh model NLP modern, termasuk BERT, GPT, dan berbagai sistem terjemahan mutakhir.

## 7. Recent Language Models: ELMo, ULMFiT, GPT, and BERT

Bagian penutup bab ini membahas **lompatan besar dalam Natural Language Processing (NLP)** pada periode 2018–2019, yang secara drastis mengubah cara representasi dan pemodelan bahasa.

### ELMo (Embeddings from Language Models)
**ELMo** memperkenalkan konsep **contextualized word embeddings**, di mana representasi suatu kata diambil dari internal state language model bidirectional yang dalam. Dengan pendekatan ini, kata yang sama dapat memiliki embedding berbeda tergantung konteks kalimatnya, sehingga mampu menangkap makna polisemik secara lebih akurat dibanding embedding statis seperti Word2Vec atau GloVe.

### ULMFiT (Universal Language Model Fine-tuning)
**ULMFiT** menunjukkan bahwa **pretraining language model berbasis LSTM** menggunakan self-supervised learning, lalu melakukan fine-tuning bertahap untuk tugas downstream, dapat menghasilkan peningkatan performa yang signifikan. Pendekatan ini sangat efektif bahkan ketika jumlah data berlabel terbatas, dan menjadi salah satu fondasi transfer learning modern di NLP.

### GPT and GPT-2
**GPT dan GPT-2** menggunakan tumpukan **Transformer decoder** besar dengan **masked multi-head attention**, dan dilatih sebagai language model autoregresif. Model ini menunjukkan kemampuan **zero-shot** dan **few-shot learning** pada berbagai tugas NLP, hanya dengan memanfaatkan pengetahuan bahasa yang dipelajari selama pretraining skala besar.

### BERT (Bidirectional Encoder Representations from Transformers)
**BERT** memperkenalkan penggunaan **Transformer encoder bidirectional** dengan dua objektif pretraining utama:
- **Masked Language Model (MLM)**, di mana sebagian token disembunyikan dan harus diprediksi dari konteks dua arah,
- **Next Sentence Prediction (NSP)**, untuk mempelajari relasi antar-kalimat.

Pendekatan ini menghasilkan **representasi bahasa universal** yang sangat kuat dan dapat di-*fine-tune* dengan ringan untuk berbagai tugas NLP seperti klasifikasi, QA, dan NLI.

---

### Key Innovations

Terobosan utama dari model-model ini meliputi:
- penggunaan **tokenisasi subword** (BPE, WordPiece, SentencePiece),
- pergeseran arsitektur dari **LSTM ke Transformer**,
- serta **self-supervised pretraining skala besar** yang kemudian diadaptasi secara efisien ke banyak tugas downstream.

Pendekatan ini kini menjadi standar de facto dalam pengembangan model bahasa modern.
